```python
### version 2
train_data, val_data, test_data = [], [], []

for maxd in range(1, tr_depth+1):
    int2letter = {i:c for i, c in enumerate(random.sample(letters, te_depth+1))}
    x= ""
    for d in range(1, maxd+1):
        x += f"# {int2letter[d-1]} + {int2letter[d]} # {int2letter[d]} - {int2letter[d-1]} "
    x += f"<sep> # {int2letter[maxd]} + <ns> # <ns> - {int2letter[maxd]}"
    x = x.split()
    val_data.append((
        x,
        ["-1"]*(len(x)-8) + x[-8:] + ["<eos>"],
    ))
for loop in range(320):
    for maxd in range(1, tr_depth+1):
        int2letter = {i:c for i, c in enumerate(random.sample(letters, te_depth+1))}
        x = ""
        for d in range(1, maxd+1):
            x += f"# {int2letter[d-1]} + {int2letter[d]} # {int2letter[d]} - {int2letter[d-1]} "
        x += f"<sep> # {int2letter[maxd]} + <ns> # <ns> - {int2letter[maxd]}"
        x = x.split()
        train_data.append((
            x,
            ["-1"]*(len(x)-8) + x[-8:] + ["<eos>"],
        ))
for loop in range(52):
    for maxd in range(tr_depth+1, te_depth+1):
        int2letter = {i:c for i, c in enumerate(random.sample(letters, te_depth+1))}
        x = ""
        for d in range(1, maxd+1):
            x += f"# {int2letter[d-1]} + {int2letter[d]} # {int2letter[d]} - {int2letter[d-1]} "
        x += f"<sep> # {int2letter[maxd]} + <ns> # <ns> - {int2letter[maxd]}"
        x = x.split()
        test_data.append((
            x,
            ["-1"]*(len(x)-8) + x[-8:] + ["<eos>"],
        ))
```

```python
### version 3
train_data, val_data, test_data = [], [], []

for maxd in range(1, tr_depth+1):
    int2letter = {i:c for i, c in enumerate(random.sample(letters, te_depth+1))}
    x = ""
    edge_list = []
    for d in range(1, maxd+1):
        edge_list.append(f"# {int2letter[d-1]} + {int2letter[d]} # {int2letter[d]} - {int2letter[d-1]} ")
    random.shuffle(edge_list)
    x += "".join(edge_list)
    x += f"<sep> # {int2letter[maxd]} + <ns> # <ns> - {int2letter[maxd]}"
    x = x.split()
    val_data.append((
        x,
        ["-1"]*(len(x)-8) + x[-8:] + ["<eos>"],
    ))
for loop in range(640):
    for maxd in range(1, tr_depth+1):
        int2letter = {i:c for i, c in enumerate(random.sample(letters, te_depth+1))}
        x = ""
        edge_list = []
        for d in range(1, maxd+1):
            edge_list.append(f"# {int2letter[d-1]} + {int2letter[d]} # {int2letter[d]} - {int2letter[d-1]} ")
        random.shuffle(edge_list)
        x += "".join(edge_list)
        x += f"<sep> # {int2letter[maxd]} + <ns> # <ns> - {int2letter[maxd]}"
        x = x.split()
        train_data.append((
            x,
            ["-1"]*(len(x)-8) + x[-8:] + ["<eos>"],
        ))
for loop in range(52):
    for maxd in range(tr_depth+1, te_depth+1):
        int2letter = {i:c for i, c in enumerate(random.sample(letters, te_depth+1))}
        x = ""
        edge_list = []
        for d in range(1, maxd+1):
            edge_list.append(f"# {int2letter[d-1]} + {int2letter[d]} # {int2letter[d]} - {int2letter[d-1]} ")
        random.shuffle(edge_list)
        x += "".join(edge_list)
        x += f"<sep> # {int2letter[maxd]} + <ns> # <ns> - {int2letter[maxd]}"
        x = x.split()
        test_data.append((
            x,
            ["-1"]*(len(x)-8) + x[-8:] + ["<eos>"],
        ))
```

```python
with open(f"/data/yingshac/model_succ/data/model_succ3_{tr_depth}_{te_depth}/lm/train.txt", "w") as f:
    for x in train_data: f.write(json.dumps(x) + "\n")
with open(f"../data/model_succ3_{tr_depth}_{te_depth}/lm/val.txt", "w") as f:
    for x in val_data: f.write(json.dumps(x) + "\n")
with open(f"../data/model_succ3_{tr_depth}_{te_depth}/lm/test.txt", "w") as f:
    for x in test_data: f.write(json.dumps(x) + "\n")
```
