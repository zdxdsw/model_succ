```python
import numpy as np
import os, json, random
from matplotlib import pyplot as plt
from collections import Counter
```

```python
def compute_entropy(counter):
    total = sum(counter.values())
    entropy = 0
    for k, v in counter.items():
        p = v / total
        entropy -= p * np.log2(p)
    return entropy
```

```python
data_dir = "../data/dyck1_1/0327_164249"
```

```python
counter_1_1 = Counter()
with open(os.path.join(data_dir, "val.txt"), "r") as f:
    lines = f.readlines()
    for l in lines:
        input_seq = "".join(json.loads(l.strip())[0])
        counter_1_1[input_seq] += 1
```

```python
print("#unique samples = ", len(counter_1_1))
print("#entropy if uniform = ", np.log2(len(counter_1_1)))
total_samples = sum(counter_1_1.values())
print("#total samples = ", total_samples)
print("entropy = ", compute_entropy(counter_1_1))
```

```python
counter_1_2 = Counter()
_1_2minus1_1 = 0
with open(os.path.join(data_dir, "val_dyck1_2.txt"), "r") as f:
    lines = f.readlines()
    for l in lines:
        input_seq = "".join(json.loads(l.strip())[0])
        if not input_seq in counter_1_1:
            _1_2minus1_1 += 1
            counter_1_2[input_seq] += 1
```

```python
print("#non-duplicating samples = ", _1_2minus1_1)
print("#unique non-duplicating samples = ", len(counter_1_2))
counter2 = counter_1_2 + counter_1_1
print("\n----- Level 2 -----")
print("#unique samples = ", len(counter2))
print("#entropy if uniform = ", np.log2(len(counter2)))
total_samples = sum(counter2.values())
print("#total samples = ", total_samples)
print("entropy = ", compute_entropy(counter2))
```

```python
counter_1_3 = Counter()
_1_3minus1_1u1_2 = 0
with open(os.path.join(data_dir, "val_dyck1_3.txt"), "r") as f:
    lines = f.readlines()
    for l in lines:
        input_seq = "".join(json.loads(l.strip())[0])
        if not input_seq in counter2:
            _1_3minus1_1u1_2 += 1
            counter_1_3[input_seq] += 1
```

```python
print("#non-duplicating samples = ", _1_3minus1_1u1_2)
print("#unique non-duplicating samples = ", len(counter_1_3))
counter3 = counter_1_1 + counter_1_2 + counter_1_3
print("\n----- Level 3 -----")
print("#unique samples = ", len(counter3))
print("#entropy if uniform = ", np.log2(len(counter3)))
total_samples = sum(counter3.values())
print("#total samples = ", total_samples)
print("entropy = ", compute_entropy(counter3))
```

```python
counter_1_4 = Counter()
_1_4minus1_1u1_2u1_3 = 0
with open(os.path.join(data_dir, "val_dyck1_4.txt"), "r") as f:
    lines = f.readlines()
    for l in lines:
        input_seq = "".join(json.loads(l.strip())[0])
        if not input_seq in counter3:
            _1_4minus1_1u1_2u1_3 += 1
            counter_1_4[input_seq] += 1
```

```python
print("#non-duplicating samples = ", _1_4minus1_1u1_2u1_3)
print("#unique non-duplicating samples = ", len(counter_1_4))
counter4 = counter_1_1 + counter_1_2 + counter_1_3 + counter_1_4
print("\n----- Level 4 -----")
print("#unique samples = ", len(counter4))
print("#entropy if uniform = ", np.log2(len(counter4)))
total_samples = sum(counter4.values())
print("#total samples = ", total_samples)
print("entropy = ", compute_entropy(counter4))
```
